```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("WineQT.csv")
print("First 3 rows:")
print(df.head(3))
print("\nLast 3 rows:")
print(df.tail(3))
```

```python
print(df.dtypes)
```

```python
print(df.describe())
```

```python
print(df.isna().sum())
```

```python
print(df.isna().any())
```

```python
mean = df["alcohol"].mean()
median = df["alcohol"].median()
mode = df["alcohol"].mode()
print("Mean =", mean)
print("Median =", median)
print("Mode =", mode)
```

```python
mean = df["quality"].mean()
median = df["quality"].median()
mode = df["quality"].mode()
print("Mean =", mean)
print("Median =", median)
print("Mode =", mode)
```

```python
print("Standard Deviation =", df["alcohol"].std())
print("Variance =", df["alcohol"].var())
print("Minimum =", df["alcohol"].min())
print("Maximum =", df["alcohol"].max())
print("Range =", df["alcohol"].max() - df["alcohol"].min())
```

```python
print(df["alcohol"].nlargest(30))
```

```python
df["alcohol"].nlargest(30).plot(kind="bar",figsize=(10, 6))
plt.title("Top 30 Highest Alcohol Values")
plt.xlabel("Wine Index")
plt.ylabel("Alcohol")
plt.show()
```

```python
sns.histplot(df["alcohol"], kde=True)
plt.title("Distribution of Alcohol")
plt.xlabel("Alcohol")
plt.ylabel("Frequency")
plt.show()
```

```python
sns.boxplot(x="alcohol", data=df)
plt.title("Box Plot of Alcohol")
plt.show()
```

```python
plt.scatter(df["alcohol"], df["quality"])
plt.title("Scatter Plot of Alcohol vs Quality")
plt.xlabel("Alcohol")
plt.ylabel("Quality")
plt.show()
```

```python
plt.scatter(df["volatile acidity"],df["quality"])
plt.title("Scatter Plot of Volatile Acidity vs Quality")
plt.xlabel("Volatile Acidity")
plt.ylabel("Quality")
plt.show()
```

```python
sns.boxplot(x="quality",y="alcohol",data=df)
plt.title("Alcohol Distribution for Different Wine Quality Levels")
plt.xlabel("Wine Quality")
plt.ylabel("Alcohol")
plt.show()
```

```python
sns.boxplot(x="quality",y="volatile acidity",data=df)
plt.title("Volatile Acidity for Different Wine Quality Levels")
plt.xlabel("Wine Quality")
plt.ylabel("Volatile Acidity")
plt.show()
```

```python
sns.pairplot(df,vars=["alcohol","pH","sulphates","volatile acidity","quality"])
plt.show()
```

```python
sns.pairplot(df,vars=["alcohol","pH","sulphates","quality"],kind="reg")
plt.show()
```

```python
sns.set(style="ticks", color_codes=True)
sns.pairplot(df,height=2,vars=["alcohol","pH","sulphates","volatile acidity"],hue="quality")
plt.show()
```

```python
from scipy import stats
corr = stats.pearsonr(df["alcohol"],df["quality"])
print("p-value:", corr[1])
print("Correlation:", corr[0])
```

```python
correlation = df.corr(method="pearson",numeric_only=True)
print(correlation)
```

```python
plt.figure(figsize=(12, 8))
sns.heatmap(correlation,annot=True,fmt=".2f",xticklabels=correlation.columns,yticklabels=correlation.columns)
plt.title("Correlation Heatmap of Wine Dataset")
plt.show()
```
